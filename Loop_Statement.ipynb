```python
odd = []
even = []

for num in range(1,21):
    if num % 2:
        even.append(num)
    else:
        odd.append(num)
print(f" Even numers {even} ")
print(f" odd numbers {odd} ")
```

```python
user_num = int(input("Enter your Number : "))
odd = []
even = []

for num in range(1,user_num+1):
    if num % 2:
        even.append(num)
    else:
        odd.append(num)
print(f" Even numers {even} ")
print(f" odd numbers {odd} ")
```

### While Loop

```python
num = 0

while num < 21:
    print(num)
    num = num+1
```

```python
num = 1
odd = []
even = []

while num < 21:
    if num % 2 == 0:
        even.append(num)
    else:
        odd.append(num)
    num += 1
print(f" Even numbers : {even} ")
print(f" odd numbers : {odd} ")
```
