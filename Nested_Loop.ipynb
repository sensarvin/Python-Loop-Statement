```python
row = int(input("Enter the row number"))

for i in range(1,row+1):
    for j in range(1,i+1):
        print("*",end="")
    print()
```

```python
row = int(input("Enter the row number"))

for i in range(row,0,-1):
    for j in range(1,i+1):
        print("*",end="")
    print()
```
